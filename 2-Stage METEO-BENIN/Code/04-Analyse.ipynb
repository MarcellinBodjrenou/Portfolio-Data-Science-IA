```python
def determin_aberante(data,variable):
    data_nouv=data.copy()
    q025=data_nouv[variable].quantile(0.25)
    q075=data_nouv[variable].quantile(0.75)
    IQR=q075-q025
    born_inf=q025-1.25*IQR
    born_sup=q075+1.25*IQR

    return data_nouv[(data_nouv[variable]<born_inf) | (born_sup<data_nouv[variable])]
```

```python
from statsmodels.tsa.stattools import adfuller

resultat_Kandi = adfuller(
    Kandi["Pluviometrie"]
)

resultat_Parakou = adfuller(
    Parakou["Pluviometrie"]
)

pd.DataFrame({"Kandi":resultat_Kandi[:2],"Parakou":resultat_Parakou[:2]},index=["ADF","P-value"])
```

```python
from statsmodels.tsa.seasonal import seasonal_decompose

decomp_Kandi = seasonal_decompose(
    Kandi["Pluviometrie"],
    model="additive",
    period=12
)
decomp_Parakou = seasonal_decompose(
    Parakou["Pluviometrie"],
    model="additive",
    period=365
)

decomp_Kandi.plot(weights=True)

decomp_Parakou.plot(weights=True) 

plt.show()
```

```python
from statsmodels.graphics.tsaplots import plot_acf

plot_acf(Kandi["Pluviometrie"],lags=30,title="ACF de la Pluviometrie de Kandi",color="red")

plt.show()
```
